```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample dataset
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Neha", "Rohit", "Pooja", "Akash"],
    "Maths": [78, 85, 67, 92, 74, 81, 88, 70],
    "Science": [82, 90, 72, 95, 76, 85, 91, 73],
    "English": [75, 88, 70, 90, 80, 78, 85, 72]
}

df = pd.DataFrame(data)

# Display dataset
print("Dataset:")
display(df)

# Summary statistics
print("Summary Statistics:")
display(df.describe())

# Calculate average marks
df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

print("Average Marks:")
display(df[["Name", "Average"]])

# 1. Bar Plot
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Name", y="Average")
plt.title("Average Marks of Students")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.show()

# 2. Histogram
plt.figure(figsize=(8,5))
sns.histplot(df["Average"], bins=5)
plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Frequency")
plt.show()

# 3. Box Plot
plt.figure(figsize=(8,5))
sns.boxplot(data=df[["Maths", "Science", "English"]])
plt.title("Box Plot of Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()

# 4. Scatter Plot
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Maths", y="Science")
plt.title("Maths vs Science Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.show()

# 5. Correlation Heatmap
plt.figure(figsize=(7,5))
sns.heatmap(
    df[["Maths", "Science", "English", "Average"]].corr(),
    annot=True
)
plt.title("Correlation Heatmap")
plt.show()
```
