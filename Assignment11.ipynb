```python
import numpy as np

# Define matrix A
A = np.array([[4, 2],
              [1, 3]])

# Define matrix B
B = np.array([10, 8])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)


# 1. Solve Matrix Equation AX = B
X = np.linalg.solve(A, B)

print("\nSolution of AX = B:")
print(X)


# 2. Find Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)


# 3. Find Inverse of Matrix A
inverse_A = np.linalg.inv(A)

print("\nInverse of Matrix A:")
print(inverse_A)


# 4. Find Transpose of Matrix A
transpose_A = np.transpose(A)

print("\nTranspose of Matrix A:")
print(transpose_A)
```
