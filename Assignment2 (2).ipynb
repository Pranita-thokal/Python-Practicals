```python
#np.arrary()
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
arr
```

```python
#np.arange()
import numpy as np
arr = np.arange(1, 7)
arr
```

```python
#np.linspace()
import numpy as np
arr = np.linspace(1, 10, 5)
arr
```

```python
#reshape()
import numpy as np
arr = np.arange(1, 7)
reshaped = arr.reshape(2, 3)
reshaped
```

```python
#Indexing
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
arr[2]
```

```python
#2D Indexing
import numpy as np
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

arr[1, 2]
```

```python
#slicing
import numpy as np
arr = np.array([10, 20, 30, 40, 50])

arr[1:4]
```

```python
#2D Slicing
import numpy as np
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

arr[:, 1:3]
```
