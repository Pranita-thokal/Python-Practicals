```python
import matplotlib.pyplot as plt

# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [24, 26, 25, 28, 30, 29, 27]

subjects = ["DS", "OS", "COA", "OOP", "Maths"]
marks = [85, 78, 90, 88, 75]

hours = [2, 3, 4, 5, 6, 7]
score = [55, 65, 70, 75, 82, 90]

scores = [55, 61, 63, 68, 70, 73, 79, 80, 81, 85, 88, 92]

# Line Plot
plt.plot(days, temperature, marker="o")
plt.title("Weekly Temperature")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.show()

# Bar Plot
plt.bar(subjects, marks)
plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()

# Scatter Plot
plt.scatter(hours, score)
plt.title("Study Hours vs Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.show()

# Histogram
plt.hist(scores, bins=6, color="orange", edgecolor="black")
plt.title("Score Distribution")
plt.xlabel("Score Range")
plt.ylabel("Frequency")
plt.show()
```
