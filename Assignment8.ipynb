```python
import matplotlib.pyplot as plt

# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [24, 26, 25, 28, 30]

subjects = ["CS", "ECE", "Math"]
scores = [85, 79, 72]

hours = [2, 3, 4, 5, 6, 7]
study_scores = [60, 65, 70, 75, 82, 90]

quiz_scores = [55, 61, 63, 68, 70, 73, 79, 80, 81, 85, 88, 92]

# Create 2 x 2 subplots
plt.subplot(2, 2, 1)
plt.plot(days, temperature, marker="o")
plt.title("Line")
plt.xlabel("Day")
plt.ylabel("Temperature")

plt.subplot(2, 2, 2)
plt.bar(subjects, scores, color="orange")
plt.title("Bar")
plt.xlabel("Subjects")
plt.ylabel("Scores")

plt.subplot(2, 2, 3)
plt.scatter(hours, study_scores)
plt.title("Scatter")
plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.subplot(2, 2, 4)
plt.hist(quiz_scores, bins=6, color="orange", edgecolor="black")
plt.title("Histogram")
plt.xlabel("Score Range")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

```
