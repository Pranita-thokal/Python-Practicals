```python
import pandas as pd
import numpy as np

# Create DataFrame with missing values
data = {
    "Name": ["Pranita", "Anjali", "Vedika", "Samiksha", "Priya"],
    "Department": ["Computer", "IT", "Computer", "IT", "Computer"],
    "Marks": [85, np.nan, 78, 92, np.nan],
    "Age": [19, 20, np.nan, 19, 20]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Check missing values
print("\nMissing values:")
print(df.isnull())

# Replace missing values
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nAfter replacing missing values:")
print(df)

# Filter rows
print("\nStudents having marks greater than 80:")
print(df[df["Marks"] > 80])

# Select columns
print("\nSelected columns:")
print(df[["Name", "Marks"]])

# Sort values
print("\nSorted by Marks:")
print(df.sort_values("Marks", ascending=False))

# Group values
print("\nAverage marks by Department:")
print(df.groupby("Department")["Marks"].mean())

# Drop missing values
print("\nData after dropna():")
print(df.dropna())
```
